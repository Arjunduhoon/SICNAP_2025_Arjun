```python
%matplotlib widget
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Extract variables
muB = df['muB']
muQ = df['muQ']
sigma = df['sigma_MF']
omega = df['omega_MF']

# Create 3D plot
fig = plt.figure(figsize=(12, 5))

# Plot 1: Sigma vs muB and muQ 
ax1 = fig.add_subplot(121, projection='3d')
sc1 = ax1.scatter(muB, sigma,muQ, c=sigma, cmap='coolwarm', s=30)
ax1.set_xlabel(r'$\mu_B$ [MeV]')
ax1.set_zlabel(r'$\mu_Q$ [MeV]')
ax1.set_ylabel(r'$\sigma_{MF}$')
ax1.set_title(r'Mean Field $\sigma$ vs $\mu_B$ and $\mu_Q$')
fig.colorbar(sc1, ax=ax1, shrink=0.5, pad=0.1)

# Plot 2: Omega vs muB and muQ
ax2 = fig.add_subplot(122, projection='3d')
sc2 = ax2.scatter(muB, omega,muQ, c=omega, cmap='viridis', s=30)
ax2.set_xlabel(r'$\mu_B$ [MeV]')
ax2.set_zlabel(r'$\mu_Q$ [MeV]')
ax2.set_ylabel(r'$\omega_{MF}$')
ax2.set_title(r'Mean Field $\omega$ vs $\mu_B$ and $\mu_Q$')
fig.colorbar(sc2, ax=ax2, shrink=0.5, pad=0.1)

plt.tight_layout()
plt.show()
plt.savefig("myplot1.png", dpi=300, bbox_inches='tight')
```

```python
%matplotlib widget
import pandas as pd
import matplotlib.pyplot as plt



# Load CSV file
df = pd.read_csv("/home/arjun/Summer_internship/Muses/Muses code with vector potential = 4/CMF_output_stable.csv",names=cmf_cols)  # Replace with your actual CSV filename


# Extract columns
E = df['E']
muQ = df['muQ']
P = df['P']

# Set up 3D figure
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# 3D Scatter Plot
scatter = ax.scatter(E, muQ, P, c=P, cmap='viridis', marker='o')

# Axis labels
ax.set_xlabel(r'Energy Density $E$ [MeV/fm$^3$]')
ax.set_ylabel(r'Charge Chemical Potential $\mu_Q$ [MeV]')
ax.set_zlabel(r'Pressure $P$ [MeV/fm$^3$]')
ax.set_title(r'3D EoS: Pressure vs Energy Density and $\mu_Q$')

# Optional color bar for pressure
cbar = plt.colorbar(scatter, ax=ax, pad=0.1)
cbar.set_label('Pressure $P$ [MeV/fm$^3$]')

plt.tight_layout()
plt.show()
```
